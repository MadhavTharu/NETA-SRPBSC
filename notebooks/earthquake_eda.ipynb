```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/NEMRC_Cleaned.csv")
df_eda = df.copy()
df_eda.sample(20)
```

```python
df_eda.info()
df_eda.describe()
```

```python
print(f'Number of records(rows) in the dataset: {df_eda.shape[0]}')
print(f'Number of features(columns) in the dataset: {df_eda.shape[1]}')
print(f'Number of duplicate entries in the dataset: {df_eda.duplicated().sum()}\n')
print(f'Number of missing values in the dataset:\n{df_eda.isna().sum()}')
```

- Here from above observations we conclude that there are 14 features including lat, long, mag, epicenter and we hae around 1245 entires with no missing values.

```python
# Magnitude Distribution of Earthquake
plt.figure(figsize=(10,7))
sns.histplot(data=df_eda['magnitude'], bins=30, kde=True, color='teal')
plt.title('Magnitude Distribution of Earthquake')
plt.show()
```

```python
# Earthquake Occurance according to District
plt.figure(figsize=(10, 30))
df_eda['epicenter'].value_counts().head(100).plot(kind='barh', color='teal')
```

```python
# Round figure Overview of Earthquake distribution by District
N = 10
count = df_eda['epicenter'].value_counts()

top = count.head(N)
others = count.iloc[N:].sum()

plotdata = top.copy()
plotdata['others']=others

plotdata.plot(kind='pie', figsize=(15, 10), startangle=90, autopct='%.1f%%')
```

```python
df_eda['datetime_utc'] = pd.to_datetime(df_eda['datetime_utc'])
df_eda['year'] = df_eda['datetime_utc'].dt.year
df_eda.groupby('year').size().plot(kind='bar', figsize=(15, 10), color='teal')
plt.title('Earthquake Occurrence by Year')
plt.ylabel('Number of Earthquakes')
plt.show()
```
